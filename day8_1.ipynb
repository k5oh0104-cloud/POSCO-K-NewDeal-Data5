```python
class person:
  #생성자 : 객체를 만드는 기능
  def __init__(self, name, age, addr, gender):
    self.name = name
    self.age = age
    self.addr = addr
    self.gender = gender

class awer:
  def __init__(self):
    print('생성자 호출!!')
```

```python
"""
생성자 constructor
객체를 생성하는 자
객체가 만들어질 때 자동으로 호출
객체가 만들어질 때 해야할 일들을 기술
객체안의 변수(인스턴스 변수) 초기화 할 때 사용
"""
class person:
  def __init__(self, name, age):
    self.name = name
    self.age = age
  def __str__(self):
    return self.name + "," + str(self.age)

p1 = person('hong', 24)
print(p1)
```

```python
class student:
  def __init__(self, name, kor, eng, math):
    self.name = name
    self.kor = kor
    self.eng = eng
    self.math = math
  def prt(self):
    print("이름: ", self.name)
    print("국어: ", self.kor)
    print("영어: ", self.eng)
    print("수학: ", self.math)
  def __str__(self):
    avg = (self.kor + self.eng + self.math)/3
    total = sum([self.kor, self.eng, self.math])
    return(self.name + "," + str(total) + "," + str(avg))

s1 = student('hong', 100, 90, 80)
print(s1)
s1.prt()
```
