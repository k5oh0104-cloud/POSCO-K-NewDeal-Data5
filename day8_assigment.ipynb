```python
import datetime
class identification:
  PID = input("주민번호를 입력하시오:\n")
  def p_information(self):
    t = self.PID.split('-')
    information = t[0]
    now = datetime.datetime.now()
    information_1 = t[1]
    year = information[:2]
    month = information[2:4]
    day = information[4:]
    # 생년월일 출력, 태어난 해 계산
    if int(information_1[0]) > 2:
      year = '20'+year
    else:
      year = '19'+year
    print(f'생년월일:{year}년 {month}월 {day}일')
    # 성별 출력
    if information_1[0] in ['1', '3']:
      print('성별:남')
    else:
      print('성별:여')
    #나이 출력
    age = now.year - int(year)
    if int(year) < now.year:
      ""
    elif int(month) < now.month:
      print('A')
      age -= 1
    elif int(day) < now.day:
      age -= 1
      print("a")
    else:
      ""
      print(" d")
    print('나이:',age)

s1 = identification()
s1.p_information()
```
