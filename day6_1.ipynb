```python
num = int(input("Please enter the number: \n"))

total = 0

for i in range (1,num+1,1):
  if i % 2 == 0:
    total -= i
  else:
    total += i

print("\n")

print("The total is: ",total)
```

```python
total = 0

data = [1, 4, 2, 6, 50, 40, 32]

new_data = [ i for i in data if i%4 == 0]

for i in new_data:
  total += i

print(f'4로 나눠지는 수들은 {new_data}입니다.')

print(f'총 합계는 {total}입니다.')
```

```python
# enumerate / zip in for loop
# enumerate = can repeat and return index no. and the value
names = ['hong', 'kim', 'lee']
for data in enumerate(names):
  print(data)
  print(data[0], data[1])
# returns tuple type
print('----------')

for index, item in enumerate(names):
  print(index, item)
#  when 'data' is printed it returns tuple type, and when requested for each items returns 'index', and 'item
```

```python
# zip matches items in list
temps = [25, 32, 28, 35, 19, 31, 27]
result = [ "이상" if i >30 else "정상" for i in temps]
print(result)

for temp, result in zip(temps, result):
  print(f'온도는 {temp}, 결과는 {result}')
```
