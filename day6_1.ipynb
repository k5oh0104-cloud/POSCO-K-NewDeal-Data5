
## Q1. Python에서 `.sort()`가 `None`을 반환하는 이유

### 1. 핵심 개념

Python의 `.sort()`는 **리스트 원본을 직접 변경(In-place Sorting)** 하는 메서드입니다.

따라서 정렬된 리스트를 별도로 반환하지 않고 `None`을 반환합니다.

이러한 설계는 원본을 변경하는 메서드와 새로운 결과를 반환하는 함수를 구분하는 데 도움이 됩니다.

### 2. 예제 코드

```python
data = [5, 2, 8, 1, 3]

result = data.sort()

print(data)    # [1, 2, 3, 5, 8]
print(result)  # None
```

### 3. `.sort()`와 `sorted()` 비교

| 구분 | `.sort()` | `sorted()` |
|---|---|---|
| 원본 변경 | O | X |
| 반환값 | None | 새로운 리스트 |
| 사용 대상 | 리스트 | 반복 가능한 객체 |
| 사용 예시 | `data.sort()` | `sorted(data)` |

### 4. 주의사항

다음과 같이 작성하면 안 됩니다.

```python
data = [5, 2, 8, 1, 3]
data = data.sort()

print(data)  # None
```

`data.sort()`의 반환값인 `None`을 다시 `data`에 할당하기 때문입니다.

### 5. 결론

- `.sort()` : 기존 리스트를 직접 정렬하고 `None` 반환
- `sorted()` : 원본을 유지하면서 새로운 정렬 리스트 반환
- 원본을 변경하는 메서드가 반드시 값을 반환하는 것은 아님

````python
"""
단어별 등장 횟수 세기

apple banana apple orange banana apple

결과

apple : 3개

banana : 2개

orange : 1개
"""
data = 'apple banana apple orange banana apple'

apple = data.split(" ").count('apple')
print(f'data중 apple의 갯수는 {apple}개입니다.')

banana = data.split(" ").count('banana')
print(f'data중 banana의 갯수는 {banana}개입니다.')

orange = data.split(" ").count('orange')
print(f'data중 orange의 갯수는 {orange}개입니다.')
````
