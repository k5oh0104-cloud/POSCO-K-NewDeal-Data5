```python
import csv
with open('08_press.csv','r', encoding = 'utf-8') as fr:
  reader = csv.reader(fr)
  header = next(reader)
  d = {}
  for i in header:
    d.update({i:[]})
  # key를 타이틀로 업데이트 완료
  for i in reader:
    for j in range(len(header)):
      d[header[j]].append(i[j])
  #value 추가 완료
print(d)
```
